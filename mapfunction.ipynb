```python
def square(x):
    return x*x
```

```python
numbers = [1,2,3,4,5,6]
list(map(square, numbers))
```

```python
## Lambda function with map
numbers = [1,2,3,4,5,6]
list(map(lambda x:x**2,numbers))
```

```python
numbers1 = [1,2,3]
numbers2 = [4,5,6]

added_numbers = list(map(lambda x,y:x+y,numbers1, numbers2))
print(added_numbers)
```

```python
## Type casting
str = ['1','2','3','4']
list(map(int,str))
```

```python
def get_name(person):
    return person['name']

people = [
    {"name":"Rajib","age":25},
    {"name": "Krish","age":26}
    ]

list(map(get_name,people))
```
